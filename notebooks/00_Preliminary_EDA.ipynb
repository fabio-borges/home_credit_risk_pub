# Preliminary Exploratory Data Analysis

```python
import os

PROJECT_DIR = ".."
DATA_DIR = f"{PROJECT_DIR}/data"
ARTIFACTS_DIR = PROJECT_DIR
```

## Generate EDA reports using ydata_profiling

```python
from ydata_profiling import ProfileReport
import pandas as pd
import home_credit_risk.data_util as data_util

def analyse(table_name):
    print(f"\n======  Analysing {table_name} ========= \n")
    csv_file = os.path.join(DATA_DIR, f"{table_name}.csv")
    df = pd.read_csv(csv_file)
    data_util.set_categorical_types(df)
    df.drop(columns=data_util.ID_COLS, inplace=True, errors='ignore')
    report_title = f"{table_name} EDA Report"
    report = ProfileReport(
        df,
        title=report_title,
        explorative=True,
        progress_bar=True,
        minimal=True)
    report_dir = os.path.join(ARTIFACTS_DIR, "eda_reports")
    os.makedirs(report_dir, exist_ok=True)
    report.to_file(os.path.join(report_dir, f"{report_title}.html"))
    print(f"Report generated as file: {report_title}.html")
```

### Table: application

```python
analyse(table_name="application_train")
```

### Table: bureau

```python
analyse("bureau")
```

### Table: bureau_balance

```python
analyse("bureau_balance")
```

```python
analyse("credit_card_balance")
```
